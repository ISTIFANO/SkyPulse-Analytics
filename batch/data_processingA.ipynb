```python
import re
import string
import emoji
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')
stop_words = set(stopwords.words('english'))

def clean_text(text):
    text = re.sub(r'<!--.*?-->', '', text, flags=re.DOTALL)
    text = emoji.replace_emoji(text, replace='')
    text = re.sub(r'@\w+', '', text)
    text = re.sub(r'http\S+|www\S+|https\S+', '', text, flags=re.MULTILINE)
    text = re.sub(r'#\w+', '', text)
    text = text.translate(str.maketrans('', '', string.punctuation))
    text = text.lower()
    words = text.split()
    words = [word for word in words if word not in stop_words]
    cleaned_text = ' '.join(words)
    return cleaned_text
data['text'] = data['text'].apply(clean_text)

```

```python
import re
import unicodedata

def text_clean(text):
    if not isinstance(text, str):
        return ""

    # Normalize unicode (removes fancy characters)
    text = unicodedata.normalize("NFKD", text)

    # Lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(r"http\S+|www\S+", "", text)

    # Remove non-letter characters (keeps spaces)
    text = re.sub(r"[^a-z\s]", " ", text)

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
import uuid
import ast
# train_df["embeddings"] = train_df["embeddings"].apply(ast.literal_eval)
# test_df["embeddings"]  = test_df["embeddings"].apply(ast.literal_eval)


# Create collections
train_collection = client.create_collection(name="my_train_collection")
test_collection  = client.create_collection(name="my_test_collection")

# Function to add in batches
def add_in_batches(collection, df, batch_size=100):
    for start_idx in range(0, len(df), batch_size):
        end_idx = start_idx + batch_size
        batch = df.iloc[start_idx:end_idx]

        collection.add(
            ids=[str(uuid.uuid4()) for _ in range(len(batch))],
            documents=batch["text"].tolist(),
            embeddings=batch["embeddings"].tolist(),
            metadatas=[{"label": int(l)} for l in batch["label"].tolist()]
        )
add_in_batches(train_collection, train_df, batch_size=1000)
add_in_batches(test_collection, test_df, batch_size=1000)
```

```python
import numpy as np

# Get embeddings and labels from ChromaDB
train_data = train_collection.get(include=["embeddings", "metadatas"])
test_data  = test_collection.get(include=["embeddings", "metadatas"])

# Convert embeddings to 2D arrays
X_train = np.array([e for e in train_data["embeddings"]])
y_train = np.array([m["label"] for m in train_data["metadatas"]])

X_test = np.array([e for e in test_data["embeddings"]])
y_test = np.array([m["label"] for m in test_data["metadatas"]])

print("X_train shape:", X_train.shape)
print("X_test  shape:", X_test.shape)
```

```python
from sklearn.model_selection import StratifiedKFold, GridSearchCV
import xgboost as xgb

param_grid = {
    "n_estimators": [200, 400],
    "max_depth": [4, 6, 8],
    "learning_rate": [0.05, 0.1],
    "subsample": [0.8, 1.0],
    "colsample_bytree": [0.8, 1.0],
    "gamma": [0, 0.1],
    "reg_alpha": [0, 0.1],
    "reg_lambda": [1, 1.5]
}

base_model = xgb.XGBClassifier(
    objective="multi:softprob",
    num_class=len(np.unique(y_train)),
    eval_metric="mlogloss",
    use_label_encoder=False,
    random_state=42
)

cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)
```

```python
from sklearn.model_selection import RandomizedSearchCV

random_search = RandomizedSearchCV(
    estimator=base_model,
    param_distributions=param_grid,
    n_iter=20,  # only 20 random combinations
    scoring="f1_macro",
    cv=3,
    verbose=2,
    n_jobs=-1,
    random_state=42
)

random_search.fit(X_train, y_train)
best_model = random_search.best_estimator_
```

```python
import numpy as np
import matplotlib.pyplot as plt
import joblib
from sklearn.metrics import (
    accuracy_score, f1_score, precision_score, recall_score,
    classification_report, confusion_matrix, roc_curve, auc
)
from sklearn.preprocessing import label_binarize
import xgboost as xgb

# -----------------------------
# Prepare data
# -----------------------------
X_train = np.array([e for e in train_data["embeddings"]])
y_train = np.array([m["label"] for m in train_data["metadatas"]])

X_test = np.array([e for e in test_data["embeddings"]])
y_test = np.array([m["label"] for m in test_data["metadatas"]])

# -----------------------------
# Train XGBoost with best hyperparameters
# -----------------------------
best_params = random_search.best_params_  # from your RandomizedSearchCV

model = xgb.XGBClassifier(
    objective="multi:softprob",
    num_class=len(np.unique(y_train)),
    eval_metric="mlogloss",
    use_label_encoder=False,
    random_state=42,
    n_jobs=-1,
    **best_params
)

model.fit(X_train, y_train)

y_pred = model.predict(X_test)
y_prob = model.predict_proba(X_test)  # probabilities for ROC

# -----------------------------
# Metrics
# -----------------------------
accuracy  = accuracy_score(y_test, y_pred)
f1        = f1_score(y_test, y_pred, average="macro")
precision = precision_score(y_test, y_pred, average="macro")
recall    = recall_score(y_test, y_pred, average="macro")

print("Accuracy :", accuracy)
print("F1-score :", f1)
print("Precision:", precision)
print("Recall   :", recall)
print("\nClassification Report:\n", classification_report(y_test, y_pred))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))

# -----------------------------
# Multi-class ROC-AUC
# -----------------------------
classes = np.unique(y_test)
y_test_bin = label_binarize(y_test, classes=classes)
n_classes = y_test_bin.shape[1]

fpr = dict()
tpr = dict()
roc_auc = dict()

for i in range(n_classes):
    fpr[i], tpr[i], _ = roc_curve(y_test_bin[:, i], y_prob[:, i])
    roc_auc[i] = auc(fpr[i], tpr[i])

# Plot ROC curves
plt.figure(figsize=(8,6))
colors = ['blue', 'green', 'red', 'orange', 'purple']

for i, color in zip(range(n_classes), colors):
    plt.plot(fpr[i], tpr[i], color=color, label=f"Class {classes[i]} (AUC={roc_auc[i]:.3f})")

plt.plot([0,1], [0,1], '--', color='gray')
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("Multi-class ROC Curve - XGBoost")
plt.legend()
plt.grid(True)
plt.show()

joblib.dump(model, "xgb_best_model.pkl")
print("Model saved as 'xgb_best_model.pkl'")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    classification_report, confusion_matrix, roc_curve, auc
)
from sklearn.preprocessing import label_binarize
import xgboost as xgb

# =========================
# Load data from ChromaDB
# =========================
train_collection = client.get_or_create_collection("my_train_collection")
test_collection  = client.get_or_create_collection("my_test_collection")

train_data = train_collection.get(include=["embeddings", "metadatas"])
test_data  = test_collection.get(include=["embeddings", "metadatas"])

def to_2d_array(embeddings):
    arr = np.array([np.array(e, dtype=float) for e in embeddings])
    dims = set(len(e) for e in arr)
    if len(dims) > 1:
        raise ValueError(f"Found inconsistent embedding dimensions: {dims}")
    return arr

X_train = to_2d_array(train_data["embeddings"])
y_train = np.array([m["label"] for m in train_data["metadatas"]])

X_test = to_2d_array(test_data["embeddings"])
y_test = np.array([m["label"] for m in test_data["metadatas"]])

print("X_train shape:", X_train.shape)
print("X_test shape :", X_test.shape)

# =========================
# Train XGBoost Classifier
# =========================
clf = xgb.XGBClassifier(
    objective='multi:softprob',  # multi-class probability
    num_class=len(np.unique(y_train)),
    eval_metric='mlogloss',
    use_label_encoder=False,
    n_estimators=500,
    learning_rate=0.1,
    max_depth=6,
    random_state=42
)

clf.fit(X_train, y_train)

# =========================
# Predictions & Probabilities
# =========================
y_pred = clf.predict(X_test)
y_prob = clf.predict_proba(X_test)  # shape = (n_samples, n_classes)


accuracy  = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average='macro')
recall    = recall_score(y_test, y_pred, average='macro')
f1        = f1_score(y_test, y_pred, average='macro')

print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1-score :", f1)

print("\nClassification Report:\n")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:\n")
print(confusion_matrix(y_test, y_pred))


classes = np.unique(y_test)
y_test_bin = label_binarize(y_test, classes=classes)
n_classes = y_test_bin.shape[1]

fpr = dict()
tpr = dict()
roc_auc = dict()

for i in range(n_classes):
    fpr[i], tpr[i], _ = roc_curve(y_test_bin[:, i], y_prob[:, i])
    roc_auc[i] = auc(fpr[i], tpr[i])

# Plot ROC curves
plt.figure(figsize=(8,6))
colors = ['blue', 'green', 'red', 'orange', 'purple']  
for i, color in zip(range(n_classes), colors):
    plt.plot(fpr[i], tpr[i], color=color, label=f"Class {classes[i]} AUC={roc_auc[i]:.4f}")
plt.plot([0,1], [0,1], '--', color='gray')
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("Multi-class ROC Curve - XGBoost")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    classification_report, confusion_matrix, roc_curve, auc
)
from sklearn.preprocessing import label_binarize, StandardScaler
from sklearn.neural_network import MLPClassifier

# =========================
# Load data from ChromaDB
# =========================
train_collection = client.get_or_create_collection("my_train_collection")
test_collection  = client.get_or_create_collection("my_test_collection")

train_data = train_collection.get(include=["embeddings", "metadatas"])
test_data  = test_collection.get(include=["embeddings", "metadatas"])

def to_2d_array(embeddings):
    arr = np.array([np.array(e, dtype=float) for e in embeddings])
    dims = set(len(e) for e in arr)
    if len(dims) > 1:
        raise ValueError(f"Found inconsistent embedding dimensions: {dims}")
    return arr

X_train = to_2d_array(train_data["embeddings"])
y_train = np.array([m["label"] for m in train_data["metadatas"]])

X_test = to_2d_array(test_data["embeddings"])
y_test = np.array([m["label"] for m in test_data["metadatas"]])

print("X_train shape:", X_train.shape)
print("X_test shape :", X_test.shape)

# =========================
# Scale features
# =========================
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled  = scaler.transform(X_test)

# =========================
#
clf = MLPClassifier(
    hidden_layer_sizes=(512, 256),
    activation='relu',
    solver='adam',
    batch_size=64,
    max_iter=500,
    random_state=42
)

clf.fit(X_train_scaled, y_train)

# =========================
# Predictions & Probabilities
# =========================
y_pred = clf.predict(X_test_scaled)
y_prob = clf.predict_proba(X_test_scaled)  # shape = (n_samples, n_classes)

# =========================
# Compute metrics
# =========================
accuracy  = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average='macro')
recall    = recall_score(y_test, y_pred, average='macro')
f1        = f1_score(y_test, y_pred, average='macro')

print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1-score :", f1)

print("\nClassification Report:\n")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:\n")
print(confusion_matrix(y_test, y_pred))

# =========================
# Multi-class ROC
# =========================
classes = np.unique(y_test)
y_test_bin = label_binarize(y_test, classes=classes)
n_classes = y_test_bin.shape[1]

fpr = dict()
tpr = dict()
roc_auc = dict()

for i in range(n_classes):
    fpr[i], tpr[i], _ = roc_curve(y_test_bin[:, i], y_prob[:, i])
    roc_auc[i] = auc(fpr[i], tpr[i])

# Plot ROC curves
plt.figure(figsize=(8,6))
colors = ['blue', 'green', 'red', 'orange', 'purple']  
for i, color in zip(range(n_classes), colors):
    plt.plot(fpr[i], tpr[i], color=color, label=f"Class {classes[i]} AUC={roc_auc[i]:.4f}")

plt.plot([0,1], [0,1], '--', color='gray')
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("Multi-class ROC Curve - MLP")
plt.legend()
plt.grid(True)
plt.show()
```
